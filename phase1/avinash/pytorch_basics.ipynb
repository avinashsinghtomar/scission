## Exercise A: tensors and indexing

```python
x=torch.arange(12).reshape(3,4)
print(x)
print(x.shape)
```

```python
#The first row of x.
print(x[0])
print(x[0].shape)
```

```python
#The last column of x.
print(x[:,3])
print(x[:,3].shape)
```

```python
#The top-left 2x2 block.
print(x[:2,:2])
print(x[:2,:2].shape)
```

```python
#The middle two columns, from all rows.
print(x[:,1:3])
print(x[:,1:3].shape)
```

```python
#Row 1, column 2, as a single number.
print(x[1,2])
print(x[1,2].shape)
```

##Exercise C: Softmax

```python
logits = torch.randn(3, 4)
probs = torch.softmax(logits, dim=-1)
print(probs, probs.sum(dim=-1))

#dim=0 → Row axis → Vertical calculation.
#dim=1 → Column axis → Horizontal calculation.
#im=-1 means the last axis, so it's equivalent to dim=1 here.
```
